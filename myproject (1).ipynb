```python
#Removing extra spaces present in the columns name
df.columns = df.columns.str.strip()
```

```python
#Dropping unneccessary columns from the dataset
df.drop("ID",axis=1,inplace=True)
df.drop("Borrower_Name",axis=1,inplace=True)
df.drop("Primary_Loan_Digit",axis=1,inplace=True)
df.drop("Borrower_State",axis=1,inplace=True)
```

```python
df.head().T
```

```python
#Removing Rs. from the columns in which it is present as because of the Rs sign numerical data is considered as object type
df['Guaranteed_Approved _Loan'] = df['Guaranteed_Approved _Loan'].replace({'Rs.': ''}, regex=True).astype(float)
df['ChargedOff_Amount'] = df['ChargedOff_Amount'].replace({'Rs.': ''}, regex=True).astype(float)
df['Loan_Approved_Gross'] = df['Loan_Approved_Gross'].replace({'Rs.': ''}, regex=True).astype(float)
df['Gross_Amount_Disbursed'] = df['Gross_Amount_Disbursed'].replace({'Rs.': ''}, regex=True).astype(float)
df['Gross_Amount_Balance'] = df['Gross_Amount_Balance'].replace({'Rs.': ''}, regex=True).astype(float)
```

```python
df.head()
```

```python
#converting the column [Date_Of_Disbursement] to date time format
col_dt=pd.to_datetime(df['Date_Of_Disbursement'], errors='coerce')
df["Date_Of_Disbursement"]=col_dt
```

```python
#converting the column [Commitment_Date] to date time format
col_dt=pd.to_datetime(df['Commitment_Date'], errors='coerce')
df["Commitment_Date"]=col_dt
```

```python
#In the column [Year_Of_Commitment] at any index , the year has been given as 1976A replacing the value
df["Year_Of_Commitment"]=df["Year_Of_Commitment"].mask(df["Year_Of_Commitment"]=="1976A",1976)
```

```python
#Converting the datatype of the column to int
df["Year_Of_Commitment"]=df["Year_Of_Commitment"].astype(int)
```

```python
df["Year_Of_Commitment"]
```

```python
#Checking null values in the dataset
df.isnull().sum()
```

```python
#Droppping null values of column Date_Of_Disbursement because we can not determine the on which date the loan is disbursed to the browwer
df = df.dropna(subset=['Date_Of_Disbursement'])
```

```python
#Relpacing null values by new
df["Business"]=df["Business"].fillna("New")
```

```python
df["Revolving_Credit_Line"] = df["Revolving_Credit_Line"].replace(["0"],['No'])
```

```python
df["Revolving_Credit_Line"].unique()
```

```python

df["Revolving_Credit_Line"] = df["Revolving_Credit_Line"].replace({
    "No":"No",
    "Yes":"Yes",
    "T":"No",
    "R":"No",
    "C":"Yes",
    "`":"None",
    ".":"None",
    "1":"Yes",
    "2":"Yes",
    "4":"Yes"
})
```

```python
df["Revolving_Credit_Line"].fillna("Unknown",inplace=True)
```

```python
df["Revolving_Credit_Line"].unique()
```

```python
#Replacing null values by no 
df["Low_Documentation_Loan"]=df["Low_Documentation_Loan"].fillna("No")
```

```python
#checking null values are treated or not 
df.isnull().sum()
```

```python
#changing date time columns into int
df['Date_Of_Disbursement'] = df['Date_Of_Disbursement'].dt.strftime('%Y%m%d').astype(int)
df['Commitment_Date'] = df['Commitment_Date'].dt.strftime('%Y%m%d').astype(int)
```

```python
#making categorical columns and numerical columns 
num_cols = ['Jobs_Reatained', 'Jobs_Created', 'Year_Of_Commitment','Guaranteed_Approved _Loan','ChargedOff_Amount','Gross_Amount_Balance',
                   'Count_Employees', 'Classification_Code','Loan_Approved_Gross','Gross_Amount_Disbursed','Loan_Term','Date_Of_Disbursement','Commitment_Date']
cat_cols= ['Business', 'Low_Documentation_Loan','Demography','Borrower_City','Code_Franchise','Name_Of_Bank','Revolving_Credit_Line','Default']
```

```python
num_df=df[num_cols]
cat_df=df[cat_cols]
```

```python
#plotting box plot for checking outliers
plt.figure(figsize=(20,25), facecolor='orange')
plotnumber =1
for column in num_df:
    if plotnumber <=15:
        ax = plt.subplot(5,3,plotnumber)
        sns.boxplot(num_df[column])
        plt.xlabel(column,fontsize=20)
    plotnumber+=1
plt.show()
```

```python
#Applying log tarsformation in the columns which does not contain any zero value
num_df[["Date_Of_Disbursement","Loan_Approved_Gross","Gross_Amount_Disbursed","Commitment_Date","Year_Of_Commitment"]]=np.log10(num_df[["Date_Of_Disbursement","Loan_Approved_Gross","Gross_Amount_Disbursed","Commitment_Date","Year_Of_Commitment"]])
```

```python
#Applying cuube rooti n the columns which contain any zero value
num_df[["Jobs_Reatained","Jobs_Created","ChargedOff_Amount","Count_Employees","Loan_Term"]]=np.cbrt(num_df[["Jobs_Reatained","Jobs_Created","ChargedOff_Amount","Count_Employees","Loan_Term"]])
```
